## Динамика продаж по месяцам

Для анализа динамики продаж данные сгруппированы по месяцу оформления заказа.

```python
sales['month_of_order'] = sales['order_purchase_timestamp'].dt.to_period('M')
```

**Количество уникальных заказов по каждому месяцу**

```python
monthly_orders = sales.groupby('month_of_order')['order_id'].nunique()
monthly_orders
```

**Стоимость проданных товаров по месяцам**

```python
monthly_sales = sales.groupby('month_of_order')['price'].sum()
monthly_sales
```

**Средний чек по месяцам**

```python
monthly_avg_check = monthly_sales / monthly_orders
monthly_avg_check
```

```python
fig = plt.figure(figsize=(10,6), dpi = 200)
axes = fig.add_axes([0,0,1,1])

axes.plot(monthly_orders.index.astype(str), monthly_orders.values, color='purple', lw = 2, marker = 'o')

axes.set_title('Количество заказов по месяцам')
axes.set_xlabel('Месяц')
axes.set_ylabel('Количество заказов')

axes.tick_params(axis='x', rotation=45)

plt.show()
```

Данные за 2016 год представлены лишь частично, поэтому основная динамика оценивается начиная с 2017 года.

В течение 2017 года наблюдается общий рост количества заказов, с заметным пиком в ноябре.

В 2018 году количество заказов остаётся относительно высоким и более стабильным.

```python
fig = plt.figure(figsize=(10,6), dpi=200)
axes = fig.add_axes([0,0,1,1])

axes.plot(monthly_avg_check.index.astype(str), monthly_avg_check.values, color='pink', lw=2, marker='o')

axes.set_title('Средний чек по месяцам')
axes.set_xlabel('Месяц')
axes.set_ylabel('Средний чек')
axes.tick_params(axis='x', rotation=45)

plt.show()
```

Средний чек в 2017–2018 годах остаётся стабильным и колеблется примерно в диапазоне 125–150. Резкое падение в декабре 2016 года связано с очень малым количеством заказов в этом месяце.

## Анализ доставки

```python
delivered_orders['delivery_days'] = (delivered_orders['order_delivered_customer_date'] - delivered_orders['order_purchase_timestamp']).dt.days
delivered_orders
```

```python
delivered_orders['delivery_days'].min()
```

```python
delivered_orders['delivery_days'].max()
```

```python
delivered_orders['delivery_days'].mean()
```

```python
delivered_orders['delivery_days'].median()
```

В среднем доставка заказа занимает около 12 дней, при этом медианный срок составляет 10 дней. Максимальный срок доставки достигает 209 дней, что указывает на наличие отдельных заказов с очень длительной доставкой.

```python
delivered_orders['delay_days'] = (delivered_orders['order_delivered_customer_date'] - delivered_orders['order_estimated_delivery_date']).dt.days
```

```python
def delivery_status(days):
    if days > 0:
        return f'Опоздание на {days} дн.'
    elif days == 0:
        return 'Доставлено в срок'
    else:
        return f'Доставлено раньше на {abs(days)} дн.'
```

```python
delivered_orders['delivery_status'] = (delivered_orders['delay_days'].apply(delivery_status))
```

```python
delivered_orders.head()
```

```python
late_orders = delivered_orders[delivered_orders['delay_days'] > 0]
late_orders
```

```python
late_orders.shape[0]
```

```python
late_orders['delay_days'].mean()
```

```python
late_orders['delay_days'].median()
```

```python
late_orders.shape[0] / delivered_orders.shape[0] * 100
```

Около 6,8% доставленных заказов были доставлены позже обещанного срока. Среди опоздавших заказов средняя задержка составила около 11 дней, медианная — 7 дней.

```python
delivered_orders['month_of_order'] = (delivered_orders['order_purchase_timestamp'].dt.to_period('M'))
```

```python
monthly_delivery = (delivered_orders.groupby('month_of_order')['delivery_days'].mean())
```

```python
fig = plt.figure(figsize=(10,6), dpi=200)
axes = fig.add_axes([0,0,1,1])

axes.plot(monthly_delivery.index.astype(str), monthly_delivery.values, color='blue', lw=2, marker='o')

axes.set_title('Средний срок доставки по месяцам')
axes.set_xlabel('Месяц')
axes.set_ylabel('Средний срок доставки')
axes.tick_params(axis='x', rotation=45)

plt.show()
```

Средний срок доставки по месяцам в рассматриваемом периоде остаётся относительно стабильным и преимущественно находится в диапазоне 10–15 дней. В отдельные месяцы наблюдаются временные увеличения продолжительности доставки, однако выраженной устойчивой тенденции к увеличению или сокращению сроков доставки не выявлено. Поскольку данные за 2016 год представлены неполно, основные выводы сформулированы на основе показателей 2017–2018 годов.

## Итоговые выводы

В рамках проекта были проанализированы данные о заказах, стоимости товаров и сроках доставки интернет-магазина Olist.

- Общая стоимость проданных товаров составила около 13,2 млн бразильских реалов, а стоимость доставки — около 2,2 млн.
- Средняя стоимость товаров в заказе составила около 137 бразильских реалов, медианная — около 86,5. Более высокое значение среднего связано с наличием отдельных дорогих заказов.
- В течение 2017 года наблюдался общий рост количества заказов с заметным пиком в ноябре. В 2018 году количество заказов оставалось относительно высоким.
- Средняя стоимость товаров в заказе в 2017–2018 годах оставалась достаточно стабильной и в большинстве месяцев находилась примерно в диапазоне 125–150 бразильских реалов.
- Средний срок доставки составил около 12 дней, медианный — 10 дней.
- Около 6,8% доставленных заказов были доставлены позже предполагаемого срока. Среди опоздавших заказов средняя задержка составила около 11 дней, медианная — 7 дней.
- Средний срок доставки по месяцам в целом оставался стабильным.

Таким образом, в рассматриваемом периоде наблюдался рост количества заказов при относительно стабильной стоимости заказа и сроках доставки. При этом отдельного внимания требуют заказы с задержкой доставки, поскольку средняя продолжительность задержки среди них составляет около 11 дней.
